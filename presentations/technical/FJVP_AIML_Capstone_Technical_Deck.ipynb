**Presenter notes (about 1 minute)**

Good morning. My capstone asks a practical question for Agribank Online: when fund transfers go live, can a model rank transactions so that fraud personnel look at the riskiest ones first?

Two points frame everything that follows. First, everything here runs on PaySim, a public synthetic dataset. There is no Agribank data in this work, so these numbers describe the method, not the Bank.

Second, the main finding is a gap. With every PaySim field, models score a perfect 1.000 PR-AUC, because the simulator leaves obvious traces. When I restrict the model to what the Bank would actually know when a customer submits a transfer, the best model reaches 0.578. That second number is the honest starting point, and most of this presentation explains how I got there and what it means for a review team.

Transition: start with the business problem.

**Presenter notes (about 1.5 minutes)**

The business driver is the 2027 plan to add InstaPay and PESONet transfers to Agribank Online. More transfers mean more exposure, and BSP expects automated, real-time monitoring and an FMS for covered products.

I framed the data science task narrowly on purpose. It is supervised binary classification on the `isFraud` label. Each transaction is scored on its own, before it completes, so the model may only use information available at that moment. The output is a probability that becomes a 0 to 100 score, and a person makes the decision.

I want to be clear that this is not an FMS. An FMS combines many controls. This is one analytical component that could feed it.

On metrics: with fraud at about 0.13 percent, accuracy is useless. A model that says "legitimate" every time is 99.87 percent accurate. PR-AUC is the main metric because it is driven by how many alerts are actually fraud. The operational view asks a different question: if a team can review k cases a day, how much fraud, and how much fraud value, ends up in that queue?

**Presenter notes (about 2 minutes)**

The pipeline runs in six steps. The scope rule keeps TRANSFER and CASH\_OUT because fraud only occurs there; in production this would be a business rule that runs before the model. The split is by time, not random, because in real use a model trains on the past and scores the future.

Imbalance is handled in training only: all training fraud plus a 10 percent sample of legitimate rows, and class weights in every model. Validation and test are never rebalanced.

Feature engineering created four families. Amount features: log amount, a high-value flag, and how far the amount sits above the typical amount for its type. Time: hour and period of day. Receiving-account history: how many transfers and how much money the account received in earlier hours. Balance reconciliation: whether balances change by exactly the amount, for example sender difference = new balance + amount - old balance.

The availability review decides what is realistic. The receiving balance after the transfer cannot be a predictor: it is only written after posting, and for interbank transfers the other bank holds it.

One honest nuance. Sender balances are technically available to Agribank. Set D drops them anyway, because in PaySim they carry the account-emptying pattern, which is very likely a simulator artifact. Also, receiving-account history in PaySim covers every account, while for interbank transfers Agribank would only see its own records, so Set D is still somewhat optimistic for interbank scoring.

**Presenter notes (about 2.5 minutes)**

Explainability first. SHAP values for the selected E6 model show three main drivers: time of day, how far the amount is above the usual amount for its type, and how much the receiving account had received before. Receiving-account history is a plausible real-world signal, since mule accounts often show unusual inflows. Time of day is the weak point: PaySim puts fraud evenly across the day while legitimate activity peaks in daytime, so the model learns that night is suspicious. That may be an artifact of the simulator.

A note on the plot: the SHAP sample adds extra fraud rows to the random sample so fraud patterns are visible. Each alert carries its top three reasons, and I want to be explicit that a reason is not evidence of fraud by the customer.

Fairness. PaySim has no customer attributes, so I cannot make any demographic fairness claim. What I can do is compare error rates across transaction groups. For type, amount and balance, alert gaps stay within the fraud-rate gaps, meaning groups get more alerts because they have more fraud. Time of day is the exception. Night-time false positives run at 37.4 times the overall rate.

The mitigation test removed the four time features and retrained. The false-alert gap by time of day almost disappears, from 0.054 to 0.003, but PR-AUC drops from 0.578 to 0.419. That is a business trade-off between protecting night-time customers from false alerts and catching more fraud. It is one experiment, not a resolution.

If asked: recall and precision "with time features" use calibrated scores, while the version without time features is uncalibrated. Before calibration the with-time recall at 150 is 45.3 percent, so the like-for-like recall drop is about 7.5 points.
